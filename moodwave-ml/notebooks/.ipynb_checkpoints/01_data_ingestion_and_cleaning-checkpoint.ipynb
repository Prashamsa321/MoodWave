# Moodwave — Phase 1: Data Ingestion, Audit, Repair, and Cleaning

**Goal:** turn the three raw Spotify sources into validated, source-specific interim tables that every later phase can trust.

This notebook deliberately keeps reusable cleaning logic in `src/moodwave/`. The notebook orchestrates the phase, displays the audit, and verifies outputs rather than hiding business rules in ad-hoc cells.

## Confirmed source decisions

- Historical analysis uses the **2000–2023** range present in the supplied playlist data.
- Blank `country` values mean **Global / World Top** and are standardized to `GLOBAL`.
- Two-letter country values are preserved as country chart codes.
- Native Spotify feature values are retained; **no scaling** happens in Phase 1.
- Malformed genre rows are repaired only when they match a known deterministic pattern; otherwise they are quarantined.

## 3. Check the important source-specific findings

```python
print("Historical scope:", summary["datasets"]["historical"]["scope"])
print("\nGenre repairs:", summary["datasets"]["genre"]["repairs"])
print("\nGenre suspicious numeric values:", summary["datasets"]["genre"]["after_cleaning"]["invalid_range_counts"])
print("\nCountry scope:", summary["datasets"]["country"]["scope"])
```

### Why the zero-value genre features are not replaced here

A small number of genre rows contain `tempo = 0`, `time_signature = 0`, or `duration_ms = 0`. Phase 1 records these values in the quality report but does not guess a correction. A later model-preparation phase can decide whether to exclude or impute them based on the needs of each experiment.

## 4. Reload the canonical Parquet outputs

Successful reload proves that downstream notebooks do not need any custom CSV repair logic.

```python
historical = pd.read_parquet(ROOT / "data" / "interim" / "historical_tracks_clean.parquet")
genre = pd.read_parquet(ROOT / "data" / "interim" / "genre_tracks_clean.parquet")
country = pd.read_parquet(ROOT / "data" / "interim" / "country_chart_clean.parquet")

print("historical:", historical.shape)
print("genre:     ", genre.shape)
print("country:   ", country.shape)
```

```python
display(historical.head(3))
display(genre.head(3))
display(country.head(3))
```

## 5. Natural-key checks

```python
checks = {
    "historical_duplicate_keys": int(historical.duplicated(["track_id", "year"]).sum()),
    "genre_duplicate_keys": int(genre.duplicated(["track_id", "track_genre"]).sum()),
    "country_duplicate_keys": int(country.duplicated(["track_id", "country", "snapshot_date", "daily_rank"]).sum()),
    "country_null_scope": int(country["country"].isna().sum()),
}
checks
```

## Phase 1 completion

Phase 1 is complete when:

- all three canonical Parquet outputs reload successfully;
- all natural-key duplicate checks above are zero;
- malformed genre records are fully accounted for in the repair/quarantine report;
- `GLOBAL` is an explicit chart scope rather than a missing country;
- raw files remain unchanged.

The next phase should consume **only** `data/interim/*.parquet`, not the raw CSVs.
