# Moodwave — Phase 2: Feature Engineering and Emotional Framework

**Goal:** convert the validated Phase 1 tables into analysis/model-ready tables with a transparent emotional framework, explicit genre ambiguity, and a reusable track catalog.

Phase 2 intentionally keeps Spotify's native feature values. Scaling will happen later *inside* model pipelines after train/test splitting.

## Phase decisions

- Historical scope remains **2000–2023**.
- Blank country values were already standardized in Phase 1 to **`GLOBAL` = World Top**.
- Mood uses a simple, reproducible valence/energy split at **0.50**.
- A track may legitimately have multiple genres; this is represented rather than hidden.
- The future mood classifier will **exclude valence and energy** because those two values directly define the mood label.

## 2. Reload the processed tables

## 3. Mood framework

The four classes are deterministic emotional quadrants:

| Valence | Energy | Mood |
|---|---|---|
| high | high | **Euphoric** |
| high | low | **Peaceful** |
| low | high | **Aggressive** |
| low | low | **Melancholic** |

For both axes, **high = value ≥ 0.50**.

```python
mood_definition = json.loads((ROOT / "config" / "mood_definition.json").read_text())
mood_definition
```

```python
for name, frame in {"historical": historical, "genre": genre, "country": country}.items():
    counts = frame["mood_label"].value_counts().reindex(["Euphoric", "Peaceful", "Aggressive", "Melancholic"])
    pct = (counts / counts.sum() * 100).round(2)
    display(pd.DataFrame({"count": counts, "percent": pct}).rename_axis(name))
```

### Why this label is not a machine-learning result

`mood_label` is an **engineered analytical label**, not a prediction. It is deliberately defined from valence and energy so charts and later dashboard components share one understandable emotional vocabulary.

In Phase 4, the mood-classification experiment will test whether the quadrant can be inferred from the *other* audio properties. Feeding valence and energy back into that classifier would only teach the model the rule used to create its target.

## 4. Genre ambiguity: single-label vs multi-label tracks

```python
track_genre_counts = genre.groupby("track_id")["track_genre"].nunique()

genre_track_summary = pd.Series({
    "unique_tracks": int(track_genre_counts.size),
    "single_genre_tracks": int((track_genre_counts == 1).sum()),
    "multi_genre_tracks": int((track_genre_counts > 1).sum()),
    "max_genres_on_one_track": int(track_genre_counts.max()),
})
genre_track_summary
```

Multi-genre rows stay available for EDA. The future primary **single-label genre classifier** will use only tracks where `is_single_genre_track == True`, which avoids giving identical tracks conflicting training labels.

## 5. Derived features

```python
display(historical[["track_name", "year", "duration_ms", "duration_min", "release_decade", "valence", "energy", "mood_label"]].head())
display(genre[["track_name", "track_genre", "genre_count", "is_single_genre_track", "mood_label"]].head())
display(country[["track_name", "country", "chart_scope_type", "daily_rank", "rank_score", "mood_label"]].head())
```

`rank_score = 51 - daily_rank` is only a presentation-friendly helper: rank 1 becomes 50 and rank 50 becomes 1. The original `daily_rank` is retained and remains the authoritative field.
